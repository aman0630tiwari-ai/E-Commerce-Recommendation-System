```python
# Import Required Libraries

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Display all columns
pd.set_option("display.max_columns", None)

print("Libraries Imported Successfully!")
```

```python
# Load Datasets

electronics = pd.read_csv("amazon_dataset/electronics.csv")
products = pd.read_csv("electronics_dataset/products.csv")
reviews = pd.read_csv("electronics_dataset/reviews.csv")

print("Datasets Loaded Successfully!")
```

```python
# Check Shape of Datasets

print("Electronics Dataset Shape :", electronics.shape)
print("Products Dataset Shape    :", products.shape)
print("Reviews Dataset Shape     :", reviews.shape)
```

```python
print("Electronics Dataset Information")
electronics.info()

print("\nProducts Dataset Information")
products.info()

print("\nReviews Dataset Information")
reviews.info()
```

```python
print("Missing Values in Electronics Dataset")
print(electronics.isnull().sum())

print("\nMissing Values in Products Dataset")
print(products.isnull().sum())

print("\nMissing Values in Reviews Dataset")
print(reviews.isnull().sum())
```

```python
# Missing Values Percentage

print("Products Dataset Missing Values (%)")
missing_products = (products.isnull().sum()/len(products))*100
print(missing_products.sort_values(ascending=False))

print("\nReviews Dataset Missing Values (%)")
missing_reviews = (reviews.isnull().sum()/len(reviews))*100
print(missing_reviews.sort_values(ascending=False))

print("\nElectronics Dataset Missing Values (%)")
missing_electronics = (electronics.isnull().sum()/len(electronics))*100
print(missing_electronics.sort_values(ascending=False))
```

```python
# Check Duplicate Records

print("Duplicate Rows in Products :", products.duplicated().sum())
print("Duplicate Rows in Reviews :", reviews.duplicated().sum())
print("Duplicate Rows in Electronics :", electronics.duplicated().sum())
```

```python
# Remove Duplicate Records

products.drop_duplicates(inplace=True)
reviews.drop_duplicates(inplace=True)
electronics.drop_duplicates(inplace=True)

print("Duplicates Removed Successfully!")
```

```python
print(products.shape)
print(reviews.shape)
print(electronics.shape)
```

```python
products["rating_stars"] = products["rating_stars"].fillna(
    products["rating_stars"].median()
)
```

```python
products["rating_count"] = (
    products["rating_count"]
    .astype(str)
    .str.replace(",", "", regex=False)
    .str.replace("ratings", "", regex=False)
    .str.replace("rating", "", regex=False)
    .str.strip()
)

products["rating_count"] = pd.to_numeric(
    products["rating_count"],
    errors="coerce"
)

products["rating_count"] = products["rating_count"].fillna(0).astype(int)
```

```python
import os

# Create processed folder if it doesn't exist
os.makedirs("processed", exist_ok=True)

# Save cleaned datasets
products.to_csv("processed/products_clean.csv", index=False)
reviews.to_csv("processed/reviews_clean.csv", index=False)
electronics.to_csv("processed/electronics_clean.csv", index=False)

print("✅ Cleaned datasets saved successfully!")
```
